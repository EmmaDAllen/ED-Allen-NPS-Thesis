```python
import random
import numpy as np
import pandas as pd
import networkx as nx

from data.real_world_networks import (
    REAL_WORLD_TRAINING_NETWORKS,
    REAL_WORLD_EVALUATION_NETWORKS,
    load_processed_real_world_network
)


def summarize_real_world_networks(num_od_samples=1000, seed=5):

    rng = random.Random(seed)

    rows = []

    all_networks = {
        **{name: "training" for name in REAL_WORLD_TRAINING_NETWORKS},
        **{name: "evaluation" for name in REAL_WORLD_EVALUATION_NETWORKS}
    }

    for network_name, split in all_networks.items():

        print(f"Analyzing {network_name}...")

        # Load processed graph
        G = load_processed_real_world_network(
            network_name,
            training=(split == "training")
        )

        n = G.number_of_nodes()
        m = G.number_of_edges()
        density = m / n

        # ------------------------------------------------------------
        # PENALTIES
        # ------------------------------------------------------------

        penalties = np.array([
            data["penalty"]
            for _, _, data in G.edges(data=True)
        ])

        # ------------------------------------------------------------
        # ALTERNATIVE PATH AVAILABILITY
        # ------------------------------------------------------------

        alternative_exists = np.array([
            data.get("alternative_exists", False)
            for _, _, data in G.edges(data=True)
        ])

        alternative_rate = alternative_exists.mean()

        # ------------------------------------------------------------
        # SAMPLE SOURCE-SINK SHORTEST-PATH HOPS
        # ------------------------------------------------------------

        nodes = list(G.nodes())

        hop_counts = []

        attempts = 0
        max_attempts = num_od_samples * 20

        while len(hop_counts) < num_od_samples and attempts < max_attempts:

            attempts += 1

            s, t = rng.sample(nodes, 2)

            try:
                hops = nx.shortest_path_length(
                    G,
                    source=s,
                    target=t
                )

                hop_counts.append(hops)

            except nx.NetworkXNoPath:
                continue

        hop_counts = np.array(hop_counts)

        # ------------------------------------------------------------
        # SUMMARY
        # ------------------------------------------------------------

        rows.append({
            "network": network_name,
            "split": split,

            "nodes": n,
            "edges": m,
            "density": density,

            "penalty_mean": penalties.mean(),
            "penalty_median": np.median(penalties),
            "penalty_std": penalties.std(),
            "penalty_min": penalties.min(),
            "penalty_95": np.percentile(penalties, 95),
            "penalty_max": penalties.max(),

            "alternative_path_rate": alternative_rate,

            "od_pairs_sampled": len(hop_counts),
            "hop_mean": hop_counts.mean(),
            "hop_median": np.median(hop_counts),
            "hop_std": hop_counts.std(),
            "hop_min": hop_counts.min(),
            "hop_95": np.percentile(hop_counts, 95),
            "hop_max": hop_counts.max()
        })

    return pd.DataFrame(rows)
```

```python
real_world_summary = summarize_real_world_networks(
    num_od_samples=1000,
    seed=5
)

real_world_summary
```
