```python
import csv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob
```

```python
# Find all 10 Tropical V2 repeat evaluation files
files = sorted(glob.glob("evaluation_results_tropical_v2_shortest_path_ood_size_v2_repeat*.csv"))

print(f"Found {len(files)} files")

for file in files:
    print(file)
```

```python
dfs = []

for file in files:
    
    df = pd.read_csv(file)

    # Extract repeat number from filename
    repeat = int(file.split("v2_repeat")[-1].split(".csv")[0])

    df["training_run"] = repeat

    dfs.append(df)

results = pd.concat(dfs, ignore_index=True)

print(results.shape)
print(results.columns.tolist())
```

```python
run_summary = (
    results
    .groupby("training_run")
    .agg(
        objective_gap=("objective_gap", "mean"),
        exact_match=("exact_match", "mean"),
        hamming_distance=("hamming_distance", "mean"),
        inference_time=("inference_time", "mean")
    )
    .reset_index()
)

run_summary
```

```python
from scipy.stats import t
import numpy as np
import pandas as pd

metrics = [
    "objective_gap",
    "exact_match",
    "hamming_distance",
    "inference_time"
]

summary = []

for metric in metrics:
    
    values = run_summary[metric]

    n = len(values)
    mean = values.mean()
    std = values.std(ddof=1)
    se = std / np.sqrt(n)

    t_crit = t.ppf(0.975, df=n - 1)

    summary.append({
        "metric": metric,
        "mean": mean,
        "std": std,
        "ci_lower": mean - t_crit * se,
        "ci_upper": mean + t_crit * se,
        "min": values.min(),
        "max": values.max()
    })

stability_summary = pd.DataFrame(summary)

stability_summary
```

```python
def add_confidence_intervals(summary, metrics, confidence_multiplier=1.96):

    for metric in metrics:
        mean_col = f"{metric}_avg"
        std_col = f"{metric}_std"

        summary[f"{metric}_se"] = (summary[std_col] / np.sqrt(summary["count"]))

        summary[f"{metric}_ci_lower"] = (summary[mean_col] - confidence_multiplier * summary[f"{metric}_se"])

        summary[f"{metric}_ci_upper"] = (summary[mean_col] + confidence_multiplier * summary[f"{metric}_se"])

    return summary
```

```python
metrics_for_ci = [
    "exact_match",
    "objective_gap",
    "hamming_distance",
    "inference_time",
    "mip_solve_time",
]

summary = add_confidence_intervals(summary, metrics_for_ci)
```

```python
summary2 = add_confidence_intervals(summary2, metrics_for_ci)
```

```python
def plot_metric_with_ci(summary, metric, xlabel="Arc-to-Node Ratio (m/n)", ylabel=None, title=None):
    
    plt.figure(figsize=(9, 6))

    for model, group in summary.groupby("model_type"):
        group = group.sort_values("density")

        line, = plt.plot(
            group["density"],
            group[f"{metric}_avg"],
            marker="o",
            label=model)

        plt.plot(
            group["density"],
            group[f"{metric}_ci_lower"],
            linestyle="--",
            color=line.get_color(),
            alpha=0.6)

        plt.plot(
            group["density"],
            group[f"{metric}_ci_upper"],
            linestyle="--",
            color=line.get_color(),
            alpha=0.6)

    plt.xlabel(xlabel)
    plt.ylabel(ylabel if ylabel else metric)
    plt.title(title if title else f"{metric} by Network Density")
    plt.legend(title="Model")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
plot_metric_with_ci(
    summary,
    metric="objective_gap",
    ylabel="Average Objective Gap",
    title="Average Objective Gap by Network Density"
)
```
